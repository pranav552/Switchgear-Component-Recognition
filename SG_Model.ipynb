```python
import os
import shutil

source_root = "/content/drive/MyDrive/NEW SWITCHGEAR"
destination = "/content/drive/MyDrive/Switchgear_Dataset"

# Delete old combined dataset if it exists
if os.path.exists(destination):
    shutil.rmtree(destination)

images_dest = os.path.join(destination, "images")
labels_dest = os.path.join(destination, "labels")

os.makedirs(images_dest, exist_ok=True)
os.makedirs(labels_dest, exist_ok=True)

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

total_images = 0
total_labels = 0

print("Copying dataset...\n")

for component in sorted(os.listdir(source_root)):

    component_path = os.path.join(source_root, component)

    if not os.path.isdir(component_path):
        continue

    # Automatically find Images/images
    image_folder = None
    for folder in ["images", "Images"]:
        path = os.path.join(component_path, folder)
        if os.path.exists(path):
            image_folder = path
            break

    # Automatically find Labels/labels
    label_folder = None
    for folder in ["labels", "Labels"]:
        path = os.path.join(component_path, folder)
        if os.path.exists(path):
            label_folder = path
            break

    if image_folder is None:
        print(f"❌ Images folder not found: {component}")
        continue

    if label_folder is None:
        print(f"❌ Labels folder not found: {component}")
        continue

    img_count = 0
    lbl_count = 0

    for image in os.listdir(image_folder):

        if image.lower().endswith(image_extensions):

            shutil.copy2(
                os.path.join(image_folder, image),
                os.path.join(images_dest, image)
            )

            img_count += 1
            total_images += 1

    for label in os.listdir(label_folder):

        if label.endswith(".txt"):

            shutil.copy2(
                os.path.join(label_folder, label),
                os.path.join(labels_dest, label)
            )

            lbl_count += 1
            total_labels += 1

    print(f"{component:<35} Images: {img_count:<3} Labels: {lbl_count}")

print("\n========================================")
print("Dataset Combined Successfully")
print("========================================")
print("Total Images :", total_images)
print("Total Labels :", total_labels)
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

dataset_path = "/content/drive/MyDrive/Switchgear_Dataset"

images_path = os.path.join(dataset_path, "images")
labels_path = os.path.join(dataset_path, "labels")

# Delete previous split if it exists
for split in ["train", "valid", "test"]:
    split_path = os.path.join(dataset_path, split)
    if os.path.exists(split_path):
        shutil.rmtree(split_path)

# Create new folders
for split in ["train", "valid", "test"]:
    os.makedirs(os.path.join(dataset_path, split, "images"))
    os.makedirs(os.path.join(dataset_path, split, "labels"))

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

images = [
    f for f in os.listdir(images_path)
    if f.lower().endswith(image_extensions)
]

# 70% Train
train_files, temp_files = train_test_split(
    images,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

# Remaining 30% → 20% Validation + 10% Test
valid_files, test_files = train_test_split(
    temp_files,
    test_size=1/3,
    random_state=42,
    shuffle=True
)

def copy_files(file_list, split):

    for image in file_list:

        shutil.copy2(
            os.path.join(images_path, image),
            os.path.join(dataset_path, split, "images", image)
        )

        label = os.path.splitext(image)[0] + ".txt"

        shutil.copy2(
            os.path.join(labels_path, label),
            os.path.join(dataset_path, split, "labels", label)
        )

copy_files(train_files, "train")
copy_files(valid_files, "valid")
copy_files(test_files, "test")

print("="*45)
print("Dataset Split Completed")
print("="*45)
print(f"Train      : {len(train_files)}")
print(f"Validation : {len(valid_files)}")
print(f"Test       : {len(test_files)}")
```
